Q5 make a bar chart of service type vs avg fare

```python
import matplotlib.pyplot as plt
```

```python
service_fare = rides.groupby("service_type")["fare"].mean()
service_fare
```

```python
plt.bar(service_fare.index, service_fare.values, color="skyblue", width=0.5)
plt.xlabel("Service Type")
plt.ylabel("Average Fare")
plt.title("Service Type vs. Average Fare")
plt.show()
```
